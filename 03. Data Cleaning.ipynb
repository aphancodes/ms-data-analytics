```python
# Importing Pandas library
import pandas as pd
```

```python
# Reading the CSV into a DataFrame, then previewing the DataFrame
df = pd.read_csv('Employee Turnover Dataset.csv')
df
```

```python
# Displaying the duplicated rows of the DataFrame
df[df.duplicated()]
```

```python
# New cleaned_df DataFrame with all 99 duplicated rows dropped
cleaned_df = df.drop_duplicates()

# Checking to make sure all duplicates were dropped
print(f"Duplicate values: {cleaned_df.duplicated().sum()}")
```

```python
# Comparing the original DataFrame to the cleaned DataFrame with duplicate rows removed
print(df.shape)
print(cleaned_df.shape)
```

```python
# Counting the amount of missing values in the DataFrame for each column
cleaned_df.isna().sum()
```

```python
# Distribution of NumCompaniesPreviouslyWorked column
cleaned_df['NumCompaniesPreviouslyWorked'].value_counts(dropna = False)
```

```python
# Distribution of unique values
cleaned_df['Age'].value_counts().sort_index(ascending = False)
```

```python
# Median imputation for the NumCompaniesPrevioulyWorked column
# Finding the median value of the number of companies previously worked by age
median_by_age = cleaned_df.groupby('Age')['NumCompaniesPreviouslyWorked'].transform('median')

# Using loc to fill missing values in the column with the corresponding median_by_age value
cleaned_df.loc[:, 'NumCompaniesPreviouslyWorked'] = cleaned_df['NumCompaniesPreviouslyWorked'].fillna(median_by_age)
```

```python
# Checking the distribution of NumCompaniesPreviouslyWorked again
cleaned_df['NumCompaniesPreviouslyWorked'].value_counts(dropna = False)
```

```python
# TextMessageOptIn distribution
cleaned_df['TextMessageOptIn'].value_counts(dropna = False)
```

```python
# Mode imputation for TextMessageOptIn column
# Finding the mode of the column; [0] returns the first and only element of the Series from using mode - Yes
mode_value = cleaned_df['TextMessageOptIn'].mode()[0]

# Then use loc to fill missing values in the column with the mode
cleaned_df.loc[:, 'TextMessageOptIn'] = cleaned_df['TextMessageOptIn'].fillna(mode_value)
```

```python
# Checking the distribution of TextMessageOptIn distribution
cleaned_df['TextMessageOptIn'].value_counts(dropna = False)
```

```python
# Distribution of AnnualProfessionalDevHrs
cleaned_df['AnnualProfessionalDevHrs'].value_counts(dropna = False)
```

```python
# Distribution of unique values
cleaned_df['JobRoleArea'].value_counts()
```

```python
# Need to standardize some of the categorical values
# Create a mapping dictionary
role_mapping = {
    'Information Technology': 'InformationTechnology', 
    'Information_Technology': 'InformationTechnology',
    'Human_Resources': 'HumanResources',
    'Human Resources': 'HumanResources'}
# replace() method
cleaned_df.loc[:, 'JobRoleArea'] = cleaned_df['JobRoleArea'].replace(role_mapping)
# Distribution of unique values to check if it worked
cleaned_df['JobRoleArea'].value_counts()
```

```python
# Median imputation for the AnnualProfessionalDevHrs column
# Calculate the median values of annual professional dev hrs by job role
job_role_medians = cleaned_df.groupby('JobRoleArea')['AnnualProfessionalDevHrs'].transform('median')
    
# Using loc to fill missing values in the column with the corresponding job_role_medians value
cleaned_df.loc[:, 'AnnualProfessionalDevHrs'] = cleaned_df['AnnualProfessionalDevHrs'].fillna(job_role_medians)
```

```python
# Checking the distribution of AnnualProfessionalDevHrs again
cleaned_df['AnnualProfessionalDevHrs'].value_counts(dropna = False)
```

Duplicated rows have been removed and missing values have been dealt with.

```python
# HourlyRate has a a trailing space, need to rename the column
cleaned_df = cleaned_df.rename(columns = {'HourlyRate ': 'HourlyRate'})
# Checking the column names again
cleaned_df.columns
```

```python
# Checking the HourlyRate column
# Summary of the data
cleaned_df['HourlyRate'].describe()
```

```python
# HourlyRate is an object that needs to be changed to a float64 because it's a numerical continuous
# Need to remove the $ before converting. 
cleaned_df['HourlyRate'] = cleaned_df['HourlyRate'].str.replace('$', '', regex = False).astype(float)
# Checking conversion
cleaned_df['HourlyRate'].dtype
```

```python
# Descriptive statistics
cleaned_df['HourlyRate'].describe()
```

```python
# Distribution of unique values
cleaned_df['HourlyRate'].value_counts().sort_index(ascending = False)
```

```python
# Checking the AnnualSalary column
# Descriptive statistics
cleaned_df['AnnualSalary'].describe()
```

```python
# Looks like AnnualSalary has some negative values - let's calculate a count of negative values
cleaned_df[cleaned_df['AnnualSalary'] < 0]['AnnualSalary'].value_counts().sum()
```

```python
# Converting negative AnnualSalary values to positive
cleaned_df['AnnualSalary'] = cleaned_df['AnnualSalary'].abs()
# Descriptive statistics
cleaned_df['AnnualSalary'].describe()
```

```python
# Min AnnualSalary and Max AnnualSalary are incorrect based off of HourlyRate and HoursWeekly values
cleaned_df['AnnualSalary'].value_counts().sort_index(ascending = True)
```

```python
# Many values are calculated incorrectly, need to make fix AnnualSalary column
# Multiply HourlyRate, HoursWeekly, and 52 weeks in a year
cleaned_df['AnnualSalary'] = round((cleaned_df['HourlyRate'] * cleaned_df['HoursWeekly'] * 52), 2)
# Descriptive statistics
cleaned_df['AnnualSalary'].describe()
```

```python
# Distribution of unique values
cleaned_df['AnnualSalary'].value_counts().sort_index()
```

AnnualSalary column has been fixed! Every salary now looks reasonable for employees at a company.

```python
# Checking DrivingCommuterDistance column
# Descriptive statistics
cleaned_df['DrivingCommuterDistance'].describe()
```

```python
# Calculating how many negative values there are in DrivingCommuterDistance
cleaned_df[cleaned_df['DrivingCommuterDistance'] < 0]['DrivingCommuterDistance'].value_counts().sum()
```

```python
# Changing negative values to positive 
cleaned_df['DrivingCommuterDistance'] = cleaned_df['DrivingCommuterDistance'].abs()
```

```python
# Identify high outliers (over 100 miles)
outliers = cleaned_df['DrivingCommuterDistance'] > 100
outliers.value_counts().sort_index(ascending = False)
```

```python
# Replace outliers with the median of non-outlier values
median_distance = cleaned_df.loc[~outliers, 'DrivingCommuterDistance'].median()
cleaned_df.loc[outliers, 'DrivingCommuterDistance'] = median_distance
# Descriptive statistics
cleaned_df['DrivingCommuterDistance'].describe()
```

```python
# Distribution of unique values
cleaned_df['DrivingCommuterDistance'].value_counts().sort_index(ascending = False)
```
