```python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder
from mlxtend.frequent_patterns import apriori, association_rules
```

```python
df = pd.read_csv('Megastore_Dataset_Task_3 3.csv')
df
```

## Encoding ##

Two ordinal variables: OrderPriority, CustomerOrderSatisfaction

Two nominal variables: Segment, PaymentMethod

```python
df['OrderPriority'].value_counts()
```

```python
# Ordinal encoding of OrderPriority
priority = [['Medium', 'High']]
priority_encoder = OrdinalEncoder(categories = priority)
df['OrderPriority_encoded'] = priority_encoder.fit_transform(df[['OrderPriority']])

# Dropping OrderPriority column
df = df.drop('OrderPriority', axis = 1)
df
```

```python
df['OrderPriority_encoded'].value_counts()
```

```python
df['CustomerOrderSatisfaction'].value_counts()
```

```python
# Ordinal encoding of CustomerOrderSatisfaction
satisfaction = [['Prefer not to answer', 'Dissatisfied', 'Very Dissatisfied', 'Satisfied', 'Very Satisfied']]
satisfaction_encoder = OrdinalEncoder(categories = satisfaction)
df['CustomerOrderSatisfaction_encoded'] = satisfaction_encoder.fit_transform(df[['CustomerOrderSatisfaction']])

# Dropping CustomerOrderSatisfaction column
df = df.drop('CustomerOrderSatisfaction', axis = 1)
df
```

```python
# One hot encoding Segment
df = pd.get_dummies(df, columns = ['Segment'])
df
```

```python
# One hot encoding PaymentMethod
df = pd.get_dummies(df, columns = ['PaymentMethod'])
df
```

```python
# One hot encoding ProductName for the market basket analysis
df = pd.get_dummies(df, columns = ['ProductName'], prefix = 'Product')
df
```

```python
# CSV of cleaned dataset before transactionalizing
df.to_csv('Cleaned_Dataset_Before_Transactionalizing.csv', index = False)
```

```python
# Selecting only product columns
columns_to_keep = ['OrderID'] + [col for col in df.columns if col.startswith('Product_')]
basket_df = df[columns_to_keep]
basket_df
```

```python
# Group by OrderID, then drop the index
basket_df = basket_df.groupby('OrderID').any()
basket_df.reset_index(drop = True, inplace = True)  # This removes the OrderID index completely
basket_df
```

```python
# Save to CSV
basket_df.to_csv('Cleaned_Dataset_Transactionalized.csv', index = False)
```

## Market Basket Analysis ##

```python
# Apriori algorithm
frequent_itemsets = apriori(basket_df, min_support = 0.01, use_colnames = True)
```

```python
# Creating the association rules (only want association rules where lift >= 1)
rules = association_rules(frequent_itemsets, metric = 'lift', min_threshold = 1)
rules.head(10)
```

```python
# Sorting by highest lift values
rules.sort_values(by = 'lift', ascending = False).head()
```

```python
# Rule 1
rule_one = rules.sort_values(by = 'lift', ascending = False).iloc[0]
print("Antecedents:", rule_one['antecedents'])
print("Consequents:", rule_one['consequents'])
print("Support:", rule_one['support'])
print("Confidence:", rule_one['confidence'])
print("Lift:", rule_one['lift'])
```

```python
# Rule 2
rule_two = rules.sort_values(by = 'lift', ascending = False).iloc[1]
print("Antecedents:", rule_two['antecedents'])
print("Consequents:", rule_two['consequents'])
print("Support:", rule_two['support'])
print("Confidence:", rule_two['confidence'])
print("Lift:", rule_two['lift'])
```
