```python
# Importing Pandas, reading the Excel file into DataFrame
import pandas as pd
df = pd.read_excel("D598 Data Set.xlsx")
```

```python
# Identifying any duplicate rows in the data set
df[df.duplicated()]
```

No duplicate rows.

```python
# Grouping all IDs by state, then running descriptive statistics for all numeric variables by state
# Stored in new DataFrame stats
stats = df.groupby('Business State').agg({
    'Total Long-term Debt': ['median', 'mean', 'min', 'max'],
    'Total Equity': ['median', 'mean', 'min', 'max'],
    'Debt to Equity': ['median', 'mean', 'min', 'max'],
    'Total Liabilities': ['median', 'mean', 'min', 'max'],
    'Total Revenue': ['median', 'mean', 'min', 'max'],
    'Profit Margin': ['median', 'mean', 'min', 'max']
})

stats
```

```python
# Filtering the DataFrame to identify all businesses with debt-to-equity ratios that are negative
df[df['Debt to Equity'] < 0]
```

```python
# Calculating debt-to-income ratio for every business

# Creating a Series
# Also checking for zero revenue (float('inf') represents positive infinity)
debt_to_income_ratio = df.apply(lambda row: row['Total Long-term Debt'] / row['Total Revenue'] 
                                if row['Total Revenue'] != 0 else float('inf'), axis = 1)

# Converting the Series to a new DataFrame with the same name
debt_to_income_ratio = debt_to_income_ratio.to_frame()

# Renaming the single column from 0 to Debt to Income
debt_to_income_ratio = debt_to_income_ratio.rename(columns = {0: 'Debt to Income'})
debt_to_income_ratio
```

```python
# Concatenating the Debt to Income DataFrame with the original DataFrame
# Assigning to a new DataFrame df3
df3 = pd.concat([df, debt_to_income_ratio], axis = 1)
df3
```

```python
# Creating new stats DataFrame with additional Debt to Income column
stats = df3.groupby('Business State').agg({
    'Total Long-term Debt': ['median', 'mean', 'min', 'max'],
    'Total Equity': ['median', 'mean', 'min', 'max'],
    'Debt to Equity': ['median', 'mean', 'min', 'max'],
    'Total Liabilities': ['median', 'mean', 'min', 'max'],
    'Total Revenue': ['median', 'mean', 'min', 'max'],
    'Profit Margin': ['median', 'mean', 'min', 'max'],
    'Debt to Income':['median', 'mean', 'min', 'max']
})
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Bar chart for Top 10 States by Average Total Long-term Debt
plt.figure(figsize = (10, 8))
top10_states = stats.sort_values(('Total Long-term Debt', 'mean'), ascending = False).head(10)
sns.barplot(x = top10_states.index, y = top10_states[('Total Long-term Debt', 'mean')])
plt.title('Top 10 States by Average Total Long-term Debt')
plt.xlabel('State')
plt.ylabel('Average Total Long-term Debt')
plt.xticks(rotation = 45)
plt.tight_layout()
plt.savefig('top10_avg_total_long-term_debt.png')
```

```python
# Bar chart for Top 10 States by Median Debt-to-Equity Ratio
plt.figure(figsize = (10, 8))
top10_states = stats.sort_values(('Debt to Equity', 'median'), ascending = False).head(10)
sns.barplot(x = top10_states.index, y = top10_states[('Debt to Equity', 'median')])
plt.title('Top 10 States by Median Debt-to-Equity Ratio')
plt.xlabel('State')
plt.ylabel('Median Debt-to-Equity Ratio')
plt.xticks(rotation = 45)
plt.tight_layout()
plt.savefig('top10_median_debt-to-equity_ratio.png')
```

```python
# New DataFrame for mean of each column of stats DataFrame
mean_metrics = pd.DataFrame({
    'Debt to Equity': stats[('Debt to Equity', 'mean')],
    'Total Long-term Debt': stats[('Total Long-term Debt', 'mean')],
    'Total Equity': stats[('Total Equity', 'mean')],
    'Total Liabilities': stats[('Total Liabilities', 'mean')],
    'Total Revenue': stats[('Total Revenue', 'mean')],
    'Profit Margin': stats[('Profit Margin', 'mean')],
    'Debt to Income': stats[('Debt to Income', 'mean')]
})
```

```python
# Heatmap of the Correlation Between State-level Financial Metrics
plt.figure(figsize = (10, 8))
correlation = mean_metrics.corr()
sns.heatmap(correlation, annot = True, cmap = 'viridis', fmt = ".2f", linewidths = 0.5)
plt.title('Correlation Between State-level Financial Metrics')
plt.savefig('correlation_between_state-level_financial_metrics.png')
```

```python
# Scatterplot showing Mean Long-term Debt vs. Mean Revenue 
# Making a new DataFrame
debt_revenue = pd.DataFrame({
    'State': stats.index,
    'Mean Long-term Debt': stats[('Total Long-term Debt', 'mean')],
    'Mean Revenue': stats[('Total Revenue', 'mean')]
})

plt.figure(figsize=(10, 8))
sns.scatterplot(
    data=debt_revenue,
    x = 'Mean Long-term Debt',
    y = 'Mean Revenue',
    s = 100  
)
plt.title("Mean Long-term Debt vs. Mean Revenue")
plt.savefig('mean_long-term_debt_mean_revenue_sp.png')
```
